```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('Amazon Sale Report.csv')
```

```python
df.isnull().sum()
```

```python
df.drop(columns='ship-country', inplace=True)
df.drop(columns='fulfilled-by', inplace=True)
df.drop(columns='index', inplace=True)
df.drop(columns='Unnamed: 22', inplace=True)
df.drop(columns='Courier Status', inplace=True)
df.drop(columns='promotion-ids', inplace=True)
```

```python
df['currency'].unique()
```

```python
df['currency']=df['currency'].fillna('INR')
```

```python
df['Amount'].isna().sum()
```

```python
df.loc[df['Status'] == 'Cancelled', 'Amount'] = 0
```

```python
df['Amount'] = df['Amount'].fillna(df['Amount'].mean())
```

```python
cols_to_fill = ['ship-city', 'ship-state']
```

```python
df[cols_to_fill] = df[cols_to_fill].fillna('Unknown')
```

```python
df['ship-postal-code'] = df['ship-postal-code'].fillna('Unknown')
```

```python
df=df.drop_duplicates()
```

```python
df[df['Amount'] < 0]
```

```python
df[df['Qty'] < 0]
```

```python
df['date_column_parsed'] = pd.to_datetime(df['Date'], errors='coerce')
```

```python
df['date_column_parsed'].isna().sum()
```

```python
df['date_column_parsed']
```

```python
df.columns
```

```python
assumed_margin = 0.25  # افترضنا هامش ربح 25%

# احسب الربح الافتراضي (25% من سعر البيع)
df['Assumed_Profit'] = df['Amount'] * assumed_margin
```

```python
df['Assumed_Profit']
```

```python
df['Status'].unique()
```

```python
df['Actual_Revenue'] = np.where(
    df['Status'] == 'Shipped - Delivered to Buyer', 
    df['Amount'], 
    0
)
```

```python
df['Actual_Revenue'].sum()
```

```python
df['Has_Promotion'] = np.where(
    df['promotion-ids'].isnull(), 
    'No Promotion', 
    'With Promotion'
)
```

```python
aov_by_promotion = df.groupby('Has_Promotion')['Actual_Revenue'].mean()
```

```python
aov_by_promotion
```

```python
categories = aov_by_promotion.index
values = aov_by_promotion.values

# إنشاء الرسم البياني
plt.figure(figsize=(8, 6))  # تحديد حجم الشكل
sns.barplot(
    x=categories, 
    y=values, 
    palette=['#FF6347', '#4CAF50'] # تحديد لوحة الألوان
)

# إضافة العناوين والتسميات
plt.title('Average Revenue by Promotion Status', fontsize=16)
plt.xlabel('Promotion Status', fontsize=12)
plt.ylabel('Average Revenue (AOV)', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# إظهار الرسم
plt.show()
```

Which month or quarter records the highest sales volume? 

```python
df['Month'] = df['date_column_parsed'].dt.month
df['Year'] = df['date_column_parsed'].dt.year
df['Quarter'] = df['date_column_parsed'].dt.quarter
```

```python
monthly_sales = df.groupby('Month')['Amount'].sum().sort_values(ascending=False)
quarterly_sales = df.groupby('Quarter')['Amount'].sum().sort_values(ascending=False)
```
